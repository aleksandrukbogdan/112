```python
#!/usr/bin/env python3
"""
Рекурсивно собирает ВСЕ текстовые файлы проекта,
кроме исключённых файлов и директорий,
и записывает их в один текстовый файл.
"""

import os
from pathlib import Path
import sys

SEPARATOR = "-" * 80

EXCLUDED_DIRS = {
    # "protocol_new_venv",
    # "templates",
    ".ipynb_checkpoints",
    ".git",
    "__pycache__",
}

EXCLUDED_FILES = {
    # "favicon.ico",
    # ".env",
    ".gitignore",
}

EXCLUDED_EXTS = {
    ".ipynb",
    ".txt",
    # ".js",
}


def is_binary(path: Path) -> bool:
    """
    Простая проверка на бинарный файл.
    Читает первые 1024 байта и ищет null-byte.
    """
    try:
        with path.open("rb") as f:
            chunk = f.read(1024)
            return b"\x00" in chunk
    except Exception:
        return True


def collect_files(source_dir: str, output_file: str):
    src = Path(source_dir).expanduser().resolve()
    out = Path(output_file).expanduser().resolve()

    if not src.exists() or not src.is_dir():
        raise ValueError(f"Папка не найдена или это не папка: {src}")

    files_to_collect = []

    for root, dirs, files in os.walk(src, topdown=True, followlinks=False):
        # Удаляем исключённые директории ДО входа в них
        dirs[:] = [d for d in dirs if d not in EXCLUDED_DIRS]

        for filename in files:
            p = Path(root) / filename

            # не включаем выходной файл
            if p.resolve() == out:
                continue

            # исключаем по расширению
            if p.suffix in EXCLUDED_EXTS:
                continue

            # исключаем конкретные файлы
            if p.name in EXCLUDED_FILES:
                continue

            # пропускаем бинарники
            if is_binary(p):
                continue

            files_to_collect.append(p)

    if not files_to_collect:
        print("Не найдено подходящих файлов.")
        return

    with out.open("w", encoding="utf-8") as fout:
        for i, p in enumerate(sorted(files_to_collect)):
            fout.write(f"# {p.resolve()}\n\n")

            try:
                text = p.read_text(encoding="utf-8")
            except UnicodeDecodeError:
                text = p.read_text(encoding="latin-1", errors="ignore")

            fout.write(text.rstrip() + "\n")

            if i != len(files_to_collect) - 1:
                fout.write("\n" + SEPARATOR + "\n\n")

    print(f"Готово. Обработано файлов: {len(files_to_collect)}. Результат: {out}")


if __name__ == "__main__":
    try:
        collect_files(".", "112.txt")
    except Exception as e:
        print("Ошибка:", e)
        sys.exit(2)
```
